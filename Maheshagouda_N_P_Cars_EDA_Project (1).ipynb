## Introduction

In this project, I am doing Exploratory Data Analysis (EDA) on a cars dataset using Python. The dataset contains information about different cars, such as their manufacturing year, fuel type, transmission, mileage, engine, power, and price.

The main aim of this project is to understand the data, find useful patterns, and study the relationship between car prices and other features. I have used graphs and charts to visualize the data and understand the results easily. I have also checked and handled missing values wherever required.


```python
#IMORTING LIBRARIES
import pandas as pd # data manipulation/analysis
import numpy as np #mathematical operations
import matplotlib.pyplot as plt # visualization
import seaborn as sns # visualization
```

```python
# Summary statistics for numerical columns
print("Numerical Data Summary:")
display(df.describe())

# Summary for categorical columns
print("Categorical Data Summary:")
display(df.describe(include="object"))
```

```python
df.isnull().sum()
```

```python
# Separate numerical and categorical columns
print(df.select_dtypes(include="number").columns)
print(df.select_dtypes(include="object").columns)
```

```python
# Check missing values in each column
print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python
# Check the data types of columns with missing values
print(df[["Mileage", "Engine", "Power", "Seats", "New_Price", "Price"]].dtypes)

# Display a few rows where Mileage is missing
print(df[df["Mileage"].isnull()][["Name", "Mileage"]])
```

```python

# Fill missing values in numeric columns
df["Seats"] = df["Seats"].fillna(df["Seats"].median())
df["Price"] = df["Price"].fillna(df["Price"].median())

# Check missing values again
print(df[["Seats", "Price", "Mileage", "Engine",
          "Power", "New_Price"]].isnull().sum())
```

```python

# Convert car specifications into numeric values
df["Mileage_Num"] = pd.to_numeric(
    df["Mileage"].str.extract(r"(\d+\.?\d*)")[0],
    errors="coerce"
)

df["Engine_Num"] = pd.to_numeric(
    df["Engine"].str.extract(r"(\d+\.?\d*)")[0],
    errors="coerce"
)

df["Power_Num"] = pd.to_numeric(
    df["Power"].str.extract(r"(\d+\.?\d*)")[0],
    errors="coerce"
)

# Display the converted columns
print(df[["Mileage_Num", "Engine_Num", "Power_Num"]].head())

# Check missing values in the new columns
print(df[["Mileage_Num", "Engine_Num", "Power_Num"]].isnull().sum())
```

```python
# Fill missing values in numeric specification columns
df["Mileage_Num"] = df["Mileage_Num"].fillna(
    df["Mileage_Num"].median()
)

df["Engine_Num"] = df["Engine_Num"].fillna(
    df["Engine_Num"].median()
)

df["Power_Num"] = df["Power_Num"].fillna(
    df["Power_Num"].median()
)

# Check missing values again
print(df[["Mileage_Num", "Engine_Num", "Power_Num"]].isnull().sum())
```

```python
# Check the dataset shape
print("Dataset shape:", df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Check missing values in all columns
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.boxplot(x=df["Price"])

plt.title("Box Plot of Car Prices")
plt.xlabel("Price (in lakhs)")

plt.tight_layout()
plt.show()
```

Observation: The box plot shows the spread of car prices and helps identify unusually high or low values. Points beyond the whiskers may represent price outliers. These values should be investigated rather than automatically removed.


```python
#Univariate analysis
sns.histplot(data=df, x="Price")
plt.title("Distribution of Car Prices")
plt.show()
```

Observation: The histogram shows how car prices are distributed across the dataset. It helps identify the price ranges in which cars are concentrated and whether the distribution is skewed toward higher prices.

```python
#Bivariate analysis
sns.scatterplot (data=df, x="Year", y="Price")
plt.title("Car Year vs Price")
plt.show()
```

Observation: The scatter plot compares a car's manufacturing year with its price. It helps examine whether newer cars tend to have higher prices than older cars. Other factors, such as brand, mileage, and condition, may also influence price.

```python
#Multivariate analysis
numeric_df = df.select_dtypes(include="number")

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Car Features Correlation")
plt.show()
```

Observation: The heatmap displays correlations between numerical car features. Values closer to +1 indicate a strong positive linear relationship, while values closer to −1 indicate a strong negative linear relationship. Correlation does not necessarily mean causation

```python
# Chart 5: Number of Cars by Transmission Type

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Transmission")

plt.title("Number of Cars by Transmission Type")
plt.xlabel("Transmission Type")
plt.ylabel("Number of Cars")

plt.tight_layout()
plt.show()
```

Observation: The bar chart compares the number of manual and automatic cars in the dataset. Manual cars are more common in this dataset, with 5,204 records compared with 2,049 automatic cars.

```python
# Chart 6: Number of Cars by Fuel Type

plt.figure(figsize=(9, 5))

sns.countplot(data=df, x="Fuel_Type")

plt.title("Number of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

Observation: The bar chart compares the number of cars using different fuel types. Diesel is the most common fuel type in this dataset, with 3,852 records
